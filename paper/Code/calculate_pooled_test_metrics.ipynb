# H1N1 + H3N2 pooled test metrics

将 H1N1 和 H3N2 的**测试预测样本直接合并**，再计算 MAE、MSE、Pearson、Spearman 与 R²。

- Adaboost、Nextflu：使用 without-name 预测。
- SerumGate：使用常规 `mean` 预测。
- `titer`、`strain`、`serum`：合并 10 个 seed 的所有测试预测。
- `season`：合并 season 39--44 的所有测试预测。

核心计算逻辑位于同目录的 `calculate_pooled_test_metrics.py`；此 notebook 用于交互式调用。

```python
from pathlib import Path
import sys

# 无论从哪里启动 Jupyter，都能导入同目录的统计脚本。
CODE_DIR = Path('/home/chenyh/workspace/fluProfiler/paper/Code')
PROJECT_ROOT = CODE_DIR.parents[1]
if str(CODE_DIR) not in sys.path:
    sys.path.insert(0, str(CODE_DIR))

import calculate_pooled_test_metrics as pooled
```

```python
# 修改这里即可切换到另一批实验结果。
RESULTS_ROOT = PROJECT_ROOT / 'results/H1H3_HA1_v1.0/20260717_164256'
RESULTS_ROOT
```

```python
# 计算 pooled / micro 指标：先拼接所有 H1N1 和 H3N2 测试样本，再计算指标。
rows = [
    {'split': split, 'model': model, **pooled.pooled_metrics(RESULTS_ROOT, model, split)}
    for split in pooled.SPLITS
    for model in pooled.MODEL_SPECS
]
rows
```

```python
# 以 DataFrame 形式展示；若环境没有 pandas，也可直接查看上一单元格的 rows。
import pandas as pd

summary = pd.DataFrame(rows)[
    ['split', 'model', 'rows', 'mae', 'mse', 'pearson', 'spearman', 'r2']
]
summary.style.format({
    'rows': '{:,.0f}',
    'mae': '{:.4f}',
    'mse': '{:.4f}',
    'pearson': '{:.4f}',
    'spearman': '{:.4f}',
    'r2': '{:.4f}',
})
```

```python
split = 'season'       # titer / strain / serum / season
model = 'SerumGate'    # Adaboost / Nextflu / SerumGate

pooled.pooled_metrics(RESULTS_ROOT, model, split)
```
