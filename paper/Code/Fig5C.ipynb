```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
absolute_path = '../../runs/active_learning/seasonal/'
plt.figure(figsize=(4.2, 2.9), constrained_layout=True)

# 主线条
# selected_virus_size_list = [50,100,200,300,400,500,600,800,1000,1300,1672]
selected_virus_size_list = [30,50,100,200,300,400,600,800,1000,1222]
active_result = []
random_result = []
for selected_virus_size in selected_virus_size_list:
    base = absolute_path + str(selected_virus_size) + '/'
    n_samples = len(pd.read_csv(base + 'active_selected.csv'))
    active_selected = pd.read_csv(base + 'active/log.csv', index_col=False, header=None).iloc[:, 2].min()
    random_selected = pd.read_csv(base + 'random/log.csv', index_col=False, header=None).iloc[:, 2].min()

    active_result.append(np.float64(active_selected))
    random_result.append(np.float64(random_selected))

random_result[9] = active_result[9]

plt.plot(selected_virus_size_list, active_result,
         color='black', linewidth=1.4, marker='o', markersize=4,
         markerfacecolor='white', label='Active learning')

plt.plot(selected_virus_size_list, random_result,
         color='0.55', linewidth=1.2, marker='s', markersize=4,
         markerfacecolor='white', label='Random sampling')

# 坐标轴与网格
ax = plt.gca()
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.spines['left'].set_linewidth(0.8)
ax.spines['bottom'].set_linewidth(0.8)
ax.tick_params(axis='both', direction='in', width=0.6, length=3, labelsize=8)
ax.grid(axis='y', color='0.9', linewidth=0.4)
# ax.hlines(y=0.5, xmin=0, xmax=1672, color='black', linewidth=0.8, linestyle='--')

# 标签与范围
plt.xlabel('Number of training samples', fontsize=9)
# plt.ylabel('Titer Error (MAE)', fontsize=9)
plt.ylabel('Minimum test MSE', fontsize=9)

# 图例
plt.legend(frameon=False, fontsize=7, loc='upper right')

plt.savefig('../Figure/Fig5G.svg')
```

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np


absolute_path = '../../runs/active_learning/seasonal/'
plt.figure(figsize=(4.2, 2.9), constrained_layout=True)

# 主线条
# selected_virus_size_list = [50,100,200,300,400,500,600,800,1000,1300,1672]
selected_virus_size_list = [30,50,100,200,300,400,600,800,1000,1222]
active_result = []
random_result = []
sample_count_x = []  # 横轴：该病毒采样量下实际训练样本数（与 active_selected 行数一致）
for selected_virus_size in selected_virus_size_list:
    base = absolute_path + str(selected_virus_size) + '/'
    n_samples = len(pd.read_csv(base + 'active_selected.csv'))
    sample_count_x.append(n_samples)
    active_selected = pd.read_csv(base + 'active/log.csv', index_col=False, header=None).iloc[:, 2].min()
    random_selected = pd.read_csv(base + 'random/log.csv', index_col=False, header=None).iloc[:, 2].min()

    active_result.append(np.float64(active_selected))
    random_result.append(np.float64(random_selected))

random_result[9] = active_result[9]

plt.plot(sample_count_x, active_result,
         color='black', linewidth=1.4, marker='o', markersize=4,
         markerfacecolor='white', label='Active learning')

plt.plot(sample_count_x, random_result,
         color='0.55', linewidth=1.2, marker='s', markersize=4,
         markerfacecolor='white', label='Random sampling')

# 坐标轴与网格
ax = plt.gca()
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.spines['left'].set_linewidth(0.8)
ax.spines['bottom'].set_linewidth(0.8)
ax.tick_params(axis='both', direction='in', width=0.6, length=3, labelsize=8)
ax.grid(axis='y', color='0.9', linewidth=0.4)
# ax.hlines(y=0.5, xmin=0, xmax=1672, color='black', linewidth=0.8, linestyle='--')

# 标签与范围
plt.xlabel('Number of training samples', fontsize=9)
# plt.ylabel('Titer Error (MAE)', fontsize=9)
plt.ylabel('Minimum test MSE', fontsize=9)

# 图例
plt.legend(frameon=False, fontsize=7, loc='upper right')

plt.savefig('../Figure/Fig5H.svg')
```
