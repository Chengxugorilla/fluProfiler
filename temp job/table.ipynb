```python
import pandas as pd
```

```python
pred = pd.read_csv('./recurrent_vaccine_predictions.csv')
wide = pred.pivot(index="seq_id_a", columns="seq_id_c", values="prediction")
wide.to_csv("recurrent_vaccine_predictions.matrix.csv")
```

```python
df = pd.read_csv('./recurrent_vaccine_predictions.matrix.csv')
id_col = df.columns[0]
values = df.drop(columns=[id_col]).apply(pd.to_numeric, errors="coerce")

sorted_df = (
    df.assign(
        _gt2_count=values.gt(2).sum(axis=1),
        _row_max=values.max(axis=1),
    )
    .sort_values(["_gt2_count", "_row_max", id_col], ascending=[True, True, True])
    .drop(columns=["_gt2_count", "_row_max"])
)
```
