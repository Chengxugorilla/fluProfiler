```python
in_fa = "H3N2_HA_epi202510-202602_origin.fasta"
out_fa = "output.fasta"

with open(in_fa, "r", encoding="utf-8") as fin, open(out_fa, "w", encoding="utf-8") as fout:
    for line in fin:
        if line.startswith(">"):
            epi = line.strip().split("|")[-1]  # EPI_ISL_20300084
            fout.write(f">{epi}\n")
        else:
            fout.write(line)
```
