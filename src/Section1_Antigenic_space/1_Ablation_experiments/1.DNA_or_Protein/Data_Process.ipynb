```python
import pandas as pd
from pathlib import Path
from sklearn.model_selection import train_test_split
```

```python
def read_fasta_dict(path):
    """
    返回 dict: {id: seq}
    保留整条 header（除了开头的 >）
    """
    records = {}
    header, seq = None, []
    with Path(path).open() as f:
        for line in f:
            line = line.strip()
            if not line:
                continue
            if line.startswith('>'):
                if header:
                    records[header] = ''.join(seq)

                # header = line[1:].strip()   # ← 不再 split()
                header = line[1:].split('|')[1]
                seq = []
            else:
                seq.append(line.upper())
    if header:
        records[header] = ''.join(seq)
    return records
```

## 1. Crick_early

```python
origin_data = pd.read_csv('/mnt/zzbnew/peixunban/chenyihao/fluProfiler_source/data/data_40/all.csv')
# 用法
H1_DNA = read_fasta_dict('/mnt/zzbnew/peixunban/chenyihao/fluProfiler_source/data/Sequence/DNA/H1.fasta')
H3_DNA = read_fasta_dict('/mnt/zzbnew/peixunban/chenyihao/fluProfiler_source/data/Sequence/DNA/H3.fasta')
HA_DNA_seqences = H1_DNA | H3_DNA

data_with_DNA = origin_data.copy()
data_with_DNA['serumHA_DNA'] = data_with_DNA['serumIslID'].map(HA_DNA_seqences)
data_with_DNA['virusHA_DNA'] = data_with_DNA['virusIslID'].map(HA_DNA_seqences)
print(data_with_DNA.shape)

group_cols = ['seq_a', 'seq_b', 'seq_c', 'seq_d', 'serumPassCat', 'virusPassCat', 'serumHA_DNA', 'virusHA_DNA']
agg_dict = {c: 'first' for c in data_with_DNA.columns if c not in group_cols}
agg_dict['label'] = 'mean'
data_with_DNA_final = data_with_DNA.groupby(group_cols).agg(agg_dict).reset_index()
print(data_with_DNA_final.shape)
```

```python
train_data, test_data = train_test_split(data_with_DNA_final, test_size=0.1, random_state=42)
```

```python
train_data.to_csv('train_data.csv')
test_data.to_csv('test_data.csv')
```
