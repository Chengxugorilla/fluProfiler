```python
import pandas as pd

def print_data(season):
    test_All = pd.read_csv(f'./test_{season}/test.csv')
    test_H1N1 = test_All[test_All['Type'] == 'H1N1']
    test_H3N2 = test_All[test_All['Type'] == 'H3N2']
    # print(f'{season} All:{test_All.shape[0]} H1N1:{test_H1N1.shape[0]} H3N2: {test_H3N2.shape[0]}')
    print(test_All.shape[0], test_H1N1.shape[0], test_H3N2.shape[0])
```

```python
print_data('2023NH')
print_data('2023SH')
print_data('2024NH')
print_data('2024SH')
print_data('2025NH')
print_data('2025SH')
```
